```python
from weather_api import get_7day_forecast # import 7 day forecast function from weather_api.py
from weather_forecast import WeatherForecast # import WeatherForecast class from weather_forecast.py
from weather_utils import plot_7_day_weather, classify_weather # import plotting and classifying functions from weather_utils.py


# coordinates for Stevens Institute of Technology
lat = 40.74400786550859
lon = -74.02491546017065

raw = get_7day_forecast(lat, lon) # retrieve raw forecast data from api.weather.gov
forecast = WeatherForecast(raw) # create a WeatherForecast object using the API data

print(forecast) # print summary of forecast

print("Average high temperature:", forecast.average_high_temp()) # print average temperature for the week

display(forecast.dataframe) # display the forecast dataframe as a table

# print rainy days
print("\nRainy days:")

for d in forecast.rainy_days(): # loop through list of rainy days
    print(d)

# print weather classification
print("\nWeather classifications:")

for d in forecast.days: # loop through forecast days  
    category = classify_weather(d) # classify each day

    # print weekday and classification
    print(d.day_name, "-", category)

# generate temperature graph using matplotlib
plot_7_day_weather(forecast)
```
